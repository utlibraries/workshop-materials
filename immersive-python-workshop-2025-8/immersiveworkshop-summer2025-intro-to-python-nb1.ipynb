## **Real-world data set: ReptTraits**

For the first case study using real data, we will be using a database on reptile traits that was described in the journal *Scientific Data.*

* Oskyrko, O., Mi, C., Meiri, S. et al. ReptTraits: a comprehensive dataset of ecological traits in reptiles. (2024). *Scientific Data* 11, 243. https://doi.org/10.1038/s41597-024-03079-5

The database is published on Figshare.

* Oskyrko, O.; Mi, C.; Meiri, S.; Du, W. (2024). ReptTraits: a comprehensive dataset of ecological traits in reptiles. *Figshare*. [Dataset]. https://doi.org/10.6084/m9.figshare.24572683.v4

````
#this codeblock is complex code for making an API call to download a Figshare dataset. It is not basic content and is not explained in detail here.

import json               #already imported in previous block
import requests           #already imported in previous block
from pathlib import Path  #works with filesystem paths

print('Retrieving ReptTraits dataset...\n')
item_id = 24572683 # unique Figshare ID for each dataset
figshare_url = 'https://api.figshare.com/v2' # endpoint for Figshare public API

file_info = [] #a blank list to hold all the file metadata

r = requests.get(figshare_url + '/articles/' + str(item_id) + '/files') # API call
file_metadata = json.loads(r.text) # retrieving metadata from response
for j in file_metadata: # add the item id to each file record (used to create subfolder for files)
    j['item_id'] = item_id
    file_info.append(j) # add the file metadata to the list

for k in file_info:
    response = requests.get(f"{figshare_url}/file/download/{k['id']}") # API call
    folder_path = f'sample_data/{k["item_id"]}' # create subfolder
    Path(folder_path).mkdir(parents=True, exist_ok=True)
    file_path = f'{folder_path}/{k["name"]}' # specify file path
    with open(file_path, 'wb') as f: # save response to folder
        f.write(response.content)
    print(f'Downloaded {k["name"]} to {file_path}')
````

### **Define A More Advanced Function That Retrieves Data**

````
# This function utilizes the requests package to make an API call to the GitHub API to retrieve information about GitHub users in different cities
querylist = []
querylist.append("location:\"austin\"&followers:>=10&repos:>=10")
querylist.append("location:\"dallas\"&followers:>=10&repos:>=10")
querylist.append("location:\"houston\"&followers:>=10&repos:>=10")
# querylist.append("location:\"topeka\"&followers:>=5&repos:>=5")
# querylist.append("location:\"miami\"&followers:>=5&repos:>=5")


def api_query(query, resultsperpage):
  requesturl = "https://api.github.com/search/users?q="+ query +"&per_page=" + str(resultsperpage) +"&limit=10"
  response = requests.get(requesturl, headers={'Content-Type': 'application/json'})
  responsejson = json.loads(response.text)

  print(query + "    " + str(responsejson))


for query in querylist:
  api_query(query, "10")

````
