## Iterate Through Key Value Pairs in a Dictionary
Each row of the JSON formatted 'results' list we generated in the previous section is something called a 'dictionary' in python

A dictionary in programming languages like python is an organized group of key-value pairs, with one 'key' being associated to a corresponding 'value'. In the printed output above where we see what is contained in the first row of results, results[0], we can see that the row is organized with a key, for example "primary breed," is paired with a value, "Domestic Longhair." We can see here that dictionaries are formatted as {'key' : 'item', 'key 2' : 'item 2', ...}

````python
pprint(results[0])
````

````python
# first, let's look at the type to see what is a list and what is a dict
print("results type :", type(results))
print("results[0] type :", type(results[0]))

# in the expression below the variable names "k" and "v" stand for key and value, but can be adjusted to alternative variable names
for k, v in results[0].items():
  if k == "animal_s_name":
    print(k + ": " + str(v))

# if you want, you can try changing that "0" in results to a different number and see what animal's name is on a different row!
````

## Looping Through Items in a List
In this section, we demonstrate how to 'loop' through items in a list. This means to automate some process to occur for each item in a list, which can be useful if you are working with large datasets.

For example, suppose we want to automate a way to identify who on the pet registry needs to renew their license by a specific date (there is a yearly renewal process) and automatically compose an email to them reminding them to do it. We can accomplish these processes with two simple 'for' loops.

````python
# specifying the date we want to check to send renewal emails for
renew_check_date = pd.Timestamp('2025-01-01')
renew_by_date = renew_check_date + pd.Timedelta(days=365) #one year after the check date

# using a for loop and an if statement to find the renewals that need to be done by a certain date
renewal_list = [] #create an empty list
for idx, date in pets_df['license_issue_date'].items():
  if date == renew_check_date:
    #print(f"\nRow index: {idx}")
    #print(pets_df.loc[idx].to_string())
    renewal_list.append(idx)

emails = [] #create another empty list
# now using a for loop to generate the text for an email for each of the pet owners reminding them to renew
for idx in renewal_list:
  text = "Hello, we are contacting you about your pet registration for your "
  text += pets_df.loc[idx]['species']
  text += " "
  text += pets_df.loc[idx]['animal_s_name']
  text += ". If you and your pet are still living in Seattle, please renew their pet registration on or before "
  text += renew_by_date.strftime("%m/%d/%Y")
  text += ". Thank you!"
  emails.append(text)

#print first email to show what it looks like
pprint(emails[0])
````

## Using a While Loop
For this example, let's pretend we are making a website where pet owners can log in to view the statuses of their registered pets! We need a way to check their log in information, and a while loop is a good way to create a password checker without causing an error that prevents them from multiple attempts!

````python
# Set the correct password
correct_password = "password123"

# Initialize an empty variable for the user's input
user_guess = ""

# Loop as long as the user's guess does not match the correct password
while user_guess != correct_password:
    # Prompt the user for input inside the loop
    user_guess = input("Enter your password: ")

    if user_guess != correct_password:
        print("Incorrect password. Try again.")

# This code runs only after the while loop condition becomes False (i.e. the password is correct)
print("Correct password! Access granted.")
````

## Define A More Advanced Function That Retrieves Data

````python
# This function utilizes the requests package to make an API call to the GitHub API to retrieve information about GitHub users in different cities
querylist = []
querylist.append("location:\"austin\"&followers:>=10&repos:>=10")
querylist.append("location:\"dallas\"&followers:>=10&repos:>=10")
querylist.append("location:\"houston\"&followers:>=10&repos:>=10")
# querylist.append("location:\"topeka\"&followers:>=5&repos:>=5")
# querylist.append("location:\"miami\"&followers:>=5&repos:>=5")


def api_query(query, resultsperpage):
  requesturl = "https://api.github.com/search/users?q="+ query +"&per_page=" + str(resultsperpage) #+"&limit=10"
  response = requests.get(requesturl, headers={'Content-Type': 'application/json'})
  responsejson = json.loads(response.text)

  print(query + "    " + str(responsejson))


for query in querylist:
  api_query(query, "10")

````

## Combining Datasets
Dataframes are very versatile and can, for example, be combined with other dataframes. The following code uses a second dataset on rental property registrations in Seattle that is also downloaded through the Socrata Open Data API (https://data.seattle.gov/Built-Environment/Rental-Property-Registration/j2xh-c7vt/about_data). We can then create summaries of how many rental properties there are in each zip code and summaries of how many pets of each kind are licensed in each zip code for comparison.

````python
# Combining dfs
## Compare pets and rental property counts by zip code

# Get data from Socrata Open Data API
## First 20000 results, returned as JSON from API / converted to Python list of
# dictionaries by sodapy.
rentals = client.get("j2xh-c7vt", limit=20000)

# Convert to df
rentals_df = pd.DataFrame.from_records(rentals)

# Only get rentals with more than 5 units (want to omit houses with single bedrooms, etc.)
## Coerce to numeric
rentals_df['rentalhousingunits'] = pd.to_numeric(rentals_df['rentalhousingunits'], errors='coerce')
rentals_df_5 = rentals_df[rentals_df['rentalhousingunits'] >= 5]

# Count by zip code and turn that output table into df
rentals_zip_counts = rentals_df_5['originalzip'].value_counts().to_frame()
# print(rentals_zip_counts.head())

# Count of original pet license df by species and zip code
species_zip_counts = pets_df[['zip_code','species']].value_counts().to_frame()
# print(species_zip_counts.head())
species_zip_counts = species_zip_counts.reset_index()

# Convert pet counts from long to wide
species_zip_counts_wide = species_zip_counts.pivot(index='zip_code', columns='species', values='count')
# print(species_zip_counts.head())
species_zip_counts_wide = species_zip_counts_wide.reset_index()

# Merge dfs
rentals_pets_df = pd.merge(rentals_zip_counts, species_zip_counts_wide, left_on='originalzip', right_on='zip_code', how='inner')
## Rename one column
rentals_pets_df = rentals_pets_df.rename(columns={'count': 'rental_count'})
## Sum species count in each zip
### Replace all NaN with 0
pet_columns = ['Cat', 'Dog', 'Goat', 'Pig']
### Coerce dog column to numeric based on nan return on sum
rentals_pets_df['pets_count'] = rentals_pets_df[pet_columns].fillna(0).sum(axis=1)
rentals_pets_df['pets_count'] = rentals_pets_df[pet_columns].sum(axis=1)

## Get ratio of a species to the rental
rentals_pets_df['cat_ratio'] = rentals_pets_df['Cat'] / rentals_pets_df['rental_count']
rentals_pets_df['dog_ratio'] = rentals_pets_df['Dog'] / rentals_pets_df['rental_count']
## Get ratio of cats to dogs
rentals_pets_df['cat_dog_ratio'] = rentals_pets_df['Cat']/rentals_pets_df['Dog']
## Get proportion of all registered cats in a given zip
rentals_pets_df['cat_proportion'] = rentals_pets_df['Cat'] / rentals_pets_df['Cat'].sum()
rentals_pets_df['dog_proportion'] = rentals_pets_df['Dog'] / rentals_pets_df['Dog'].sum()
## Sort by ratio of cats to dogs, descending
rentals_pets_df = rentals_pets_df.sort_values(by='cat_dog_ratio', ascending=False)
print(rentals_pets_df.head())
````

````python
# Maybe we want to look at the frequency of commonly banned breeds in high-rental-density areas
## get df of just pitbulls
pitbulls = pets_df[pets_df['primary_breed'].str.contains('Pit Bull', case=True)]
# print(pitbulls.head())

# Count by zip code and turn that output table into df
pitbulls_df = pitbulls['zip_code'].value_counts().to_frame()
print(pitbulls_df.head())
pitbulls_df = pitbulls_df.reset_index()

## Merge with rental property count df
rentals_pits_df = pd.merge(rentals_zip_counts, pitbulls_df, left_on='originalzip', right_on='zip_code', how='inner')
## Rename two columns
rentals_pits_df = rentals_pits_df.rename(columns={'count_x': 'rental_count', 'count_y': 'pitbull_count'})
## Sort by pitbull count in descending order
rentals_pits_df = rentals_pits_df.sort_values(by='pitbull_count', ascending=False)
print(rentals_pits_df.head())
````
